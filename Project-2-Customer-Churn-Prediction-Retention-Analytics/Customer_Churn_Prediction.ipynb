```python
import kagglehub
import os
import shutil
```

```python
project_path = r"E:\Data Analyst Internship\Customer Churn Prediction & Retention Analytics"

dataset_path = os.path.join(project_path, "Dataset")

print("Project path:")
print(project_path)

print("\nDataset path:")
print(dataset_path)
```

```python
os.makedirs(dataset_path, exist_ok=True)
print("Dataset folder is ready!")
```

```python
download_path = kagglehub.dataset_download(
    "blastchar/telco-customer-churn"
)

print("Kaggle dataset downloaded.")
print("Temporary download location:")
print(download_path)
```

```python
csv_files = []

for root, dirs, files in os.walk(download_path):
    for file in files:
        if file.lower().endswith(".csv"):
            csv_files.append(os.path.join(root, file))

print("CSV files found:")

for file in csv_files:
    print(file)
```

```python
source_file = csv_files[0]

destination_file = os.path.join(
    dataset_path,
    "telco_customer_churn.csv"
)

shutil.copy2(source_file, destination_file)

print("Dataset copied successfully!")
print()
print("Dataset saved at:")
print(destination_file)
```
